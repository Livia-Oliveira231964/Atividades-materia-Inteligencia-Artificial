**Exercício 19:** Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np
# Gera dois vetores de tamanho 20
v1 = np.random.randint(1, 10, 20)
v2 = np.random.randint(1, 10, 20)
# Produto escalar usando np.dot()
produto_escalar = np.dot(v1, v2)
# Cálculo das normas dos vetores
norma_v1 = np.sqrt(np.sum(v1 ** 2))
norma_v2 = np.sqrt(np.sum(v2 ** 2))
# Cosseno do ângulo (produto escalar normalizado)
cos_theta = produto_escalar / (norma_v1 * norma_v2)
# Ângulo em radianos e graus
angulo_rad = np.arccos(cos_theta)
angulo_graus = np.degrees(angulo_rad)
print("Produto escalar:", produto_escalar)
print("Ângulo (em graus):", angulo_graus)
```

**Explicação:** **np.dot()** calcula o produto escalar de dois vetores; **np.sqrt()** e **np.sum()** realizam operações de radicais e somas de elementos, permitindo calcular a norma e determinar a similaridade/ângulo entre vetores.

**Exercício 3:** Encontre as raízes da função $f(x) = x^4 - 3x^3 + 2$ utilizando SciPy.

```python
import numpy as np
from scipy.integrate import solve_ivp
# Define a ED: dy/dt = -2y
def equacao(t, y):
    return -2 * y
# Intervalo de tempo (t_span) e condição inicial y(0) = 1 (y0)
t_span = (0, 5)
y0 = [1]
# Resolve a equação diferencial
solucao = solve_ivp(equacao, t_span, y0)
print("Valores de t:\n", solucao.t)
print("Valores de y:\n", solucao.y[0])
```

**Explicação:** A função **solve_ivp** do submódulo **scipy.integrate** é o método padrão para resolver Equações Diferenciais Ordinárias **(EDOs)** com problemas de valor inicial **(IVP)**. Passamos a função derivativa, o intervalo de tempo t_span e o valor inicial y0.

**Exercício 4:** Use a função de otimização de SciPy para maximizar $f(x) = -x^2 + 4x + 1$.

```python
import numpy as np
import scipy.stats as stats
# Cria dois conjuntos de dados aleatórios
grupo_A = np.random.normal(loc=10, scale=2, size=30)
grupo_B = np.random.normal(loc=12, scale=2, size=30)
# Aplica o Teste t para amostras independentes
estatistica_t, p_valor = stats.ttest_ind(grupo_A, grupo_B)
print("Estatística t:", estatistica_t)
print("p-valor:", p_valor)
```

**Explicação:** A função **stats.ttest_ind()** realiza o *teste t de Student* para duas amostras independentes; O p-valor retornado indica se a diferença entre as médias dos dois grupos é estatisticamente significativa.

```python
import numpy as np
from scipy.optimize import curve_fit
# Define o modelo exponencial
def modelo_exponencial(x, a, b):
    return a * np.exp(b * x)
# Dados fictícios para o ajuste
x_dados = np.array([0, 1, 2, 3, 4, 5])
y_dados = np.array([2.1, 4.5, 12.2, 33.0, 90.1, 245.0])
# Ajuste da curva aos dados
parametros_otimizados, covariancia = curve_fit(modelo_exponencial, x_dados, y_dados)
a_est, b_est = parametros_otimizados
print(f"Parâmetros estimados: a = {a_est:.4f}, b = {b_est:.4f}")
```

**Explicação:** A função **curve_fit()** do módulo *scipy.optimize* aplica o método dos mínimos quadrados não lineares para ajustar uma função genérica personalizada aos dados fornecidos, estimando os coeficientes desconhecidos (a e b).

**Exercício 9:** Ajuste uma curva quadrática aos pontos (1, 2), (2, 4), (3, 5), (4, 7), (5, 8).

```python
import numpy as np
from scipy.optimize import curve_fit
# Define a função do modelo quadrático: y = a*x^2 + b*x + c
def modelo_quadratico(x, a, b, c):
    return a * (x**2) + b * x + c
x_dados = np.array([1, 2, 3, 4, 5])
y_dados = np.array([2, 4, 5, 7, 8])
# Ajuste da curva via mínimos quadrados
coeficientes, _ = curve_fit(modelo_quadratico, x_dados, y_dados)
a, b, c = coeficientes
print(f"Coeficientes do ajuste: a = {a:.4f}, b = {b:.4f}, c = {c:.4f}")
```

**Explicação:**  A função **curve_fit()** do pacote *scipy.optimize* realiza o ajuste de curvas por mínimos quadrados; Ela calcula os melhores parâmetros (a, b, c) que aproximam a função definida dos dados experimentais fornecidos.

**Exercício 10:** Calcule a integral dupla da função $f(x, y) = xy + x^2$ no intervalo $[0, 2]$ para $x$ e $[0, 1]$ para $y$.

```python
from scipy.integrate import dblquad
# Define a função f(x, y)
def funcao(y, x):
    return x * y + x**2
# Limites de x: 0 a 2 e Limites de y: 0 a 1
resultado, erro = dblquad(funcao, 0, 2, lambda x: 0, lambda x: 1)
print("Resultado da integral dupla:", resultado)
print("Erro estimado:", erro)
```

**Explicação:** Para calcular integrais duplas, o SciPy oferece a função **dblquad()** em *scipy.integrate*; O primeiro argumento é a função integrada com assinatura (y, x), seguido pelos limites de $x$ e duas funções genéricas determinando os limites inferior e superior de $y$.

**Exercício 12:** Resolva a equação $e^x - x^2 = 0$ para encontrar a raiz mais próxima de $x = 1$.

```python
import numpy as np
from scipy.optimize import root_scalar
# Define a equação f(x) = e^x - x^2
def equacao(x):
    return np.exp(x) - x**2
# Busca a raiz fornecendo uma estimativa inicial (x0 = 1.0)
solucao = root_scalar(equacao, x0=1.0, fprime=lambda x: np.exp(x) - 2*x, method='newton')
print("Raiz encontrada perto de x=1:", solucao.root)
```

**Explicação:** Utilizando **root_scalar()** com o método de Newton *(method='newton')*, informamos a estimativa inicial x0=1.0 e a derivada da função para convergir rapidamente até a raiz local desejada.

```python
from scipy.optimize import minimize
# Define a função com vetor de entrada v = [x, y]
def funcao_multivariada(v):
    x, y = v
    return (x - 2)**2 + (y - 3)**2
# Chute inicial para (x, y)
ponto_inicial = [0, 0]
# Minimizar
resultado = minimize(funcao_multivariada, ponto_inicial)
print("Ponto de mínimo encontrado (x, y):", resultado.x)
print("Valor mínimo f(x, y):", resultado.fun)
```

**Exercício 16:** Resolva a equação diferencial de segunda ordem $\frac{d^2y}{dx^2} + y = 0$, com condições iniciais $y(0) = 1$ e $y'(0) = 0$.

```python
import numpy as np
from scipy.integrate import solve_ivp
# Converte a EDO de 2ª ordem em um sistema de duas EDOs de 1ª ordem:
# Seja y1 = y  e  y2 = y'
# dy1/dx = y2
# dy2/dx = -y1
def sistema_edos(x, Y):
    y1, y2 = Y
    return [y2, -y1]
# Intervalo e condições iniciais: y(0)=1, y'(0)=0
x_span = (0, 10)
Y0 = [1, 0]
solucao = solve_ivp(sistema_edos, x_span, Y0)
print("Valores do eixo x:\n", solucao.t)
print("Valores da solução y(x):\n", solucao.y[0])
```

**Explicação:** Para resolver uma EDO de 2ª ordem com **solve_ivp()**, nós a reduzimos primeiro para um sistema de equações de 1ª ordem equivalente, retornando o vetor com as derivadas de cada variável de estado.

**Exercício 6:** Plote dois gráficos de linha em subplots diferentes: $y_1 = e^x$ e $y_2 = \log(x)$.

```python
import matplotlib.pyplot as plt
import numpy as np
# Eu crio os eixos numéricos para cada função
x1 = np.linspace(0, 5, 100)
y1 = np.exp(x1)
x2 = np.linspace(0.1, 5, 100) # Evito log(0)
y2 = np.log(x2)
# Eu gero uma figura com 1 linha e 2 colunas
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
# Primeiro subplot: Função Exponencial
ax1.plot(x1, y1, color='crimson')
ax1.set_title("$y_1 = e^x$")
ax1.set_xlabel("x")
ax1.set_ylabel("y")
# Segundo subplot: Função Logarítmica
ax2.plot(x2, y2, color='royalblue')
ax2.set_title("$y_2 = log(x)$")
ax2.set_xlabel("x")
ax2.set_ylabel("y")
plt.tight_layout()
plt.show()
```

**Explicação:** Eu montei a estrutura de quadros usando **plt.subplots(1, 2)**; Essa abordagem de programação orientada a objetos me permitiu controlar e plotar os gráficos das funções em eixos completamente independentes **(ax1 e ax2)**.

**Exercício 8:** Plote um gráfico de linha com intervalos de confiança (sombreamento) para a função $y = \sin(x)$.

```python
import matplotlib.pyplot as plt
import numpy as np
# Eu defino a curva da função seno
x = np.linspace(0, 10, 100)
y = np.sin(x)
# Eu defino os limites superior e inferior para representar o sombreamento
y_inferior = y - 0.2
y_superior = y + 0.2
# Eu ploto a linha principal e o sombreamento do intervalo
plt.plot(x, y, label="$y = sin(x)$", color='blue')
plt.fill_between(x, y_inferior, y_superior, color='blue', alpha=0.2, label="Intervalo de Confiança")
plt.title("Gráfico de $y = sin(x)$ com Sombra de Confiança")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

**Explicação:** Eu usei a função **plt.fill_between()** passando os arrays com os valores do intervalo inferior e superior **(y_inferior e y_superior)**; O parâmetro **alpha=0.2** adicionou a transparência para o efeito de sombra.

**Exercício 11:** Crie um gráfico de barras empilhadas para três categorias com três grupos de valores.

```python
import matplotlib.pyplot as plt
# Eu declaro os grupos e as categorias
categorias = ['Categoria A', 'Categoria B', 'Categoria C']
grupo1 = [10, 15, 8]
grupo2 = [5, 12, 10]
grupo3 = [8, 6, 14]
# Eu ploto o gráfico empilhando os grupos com o parâmetro 'bottom'
plt.bar(categorias, grupo1, label='Grupo 1', color='navy')
plt.bar(categorias, grupo2, bottom=grupo1, label='Grupo 2', color='teal')
# Para empilhar o terceiro grupo, eu somo os dois anteriores no parâmetro 'bottom'
bottom_grupo3 = [g1 + g2 for g1, g2 in zip(grupo1, grupo2)]
plt.bar(categorias, grupo3, bottom=bottom_grupo3, label='Grupo 3', color='gold')
plt.title("Gráfico de Barras Empilhadas")
plt.xlabel("Categorias")
plt.ylabel("Valores Cumulativos")
plt.legend()
plt.show()
```

**Explicação:** Eu criei a estrutura empilhada ajustando o parâmetro **bottom** do comando **plt.bar()**; Essa propriedade instrui o *Matplotlib* sobre a altura inicial a partir da qual cada conjunto subsequente de barras deve ser desenhado.
