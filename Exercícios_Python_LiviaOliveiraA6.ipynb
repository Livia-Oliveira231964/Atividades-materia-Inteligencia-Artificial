**Exercício 21:** Implemente a busca binária para encontrar um elemento em uma lista ordenada.

```python
# Fonte: https://www.geeksforgeeks.org/python/python-program-for-binary-search/
# 1. Função que realiza a busca binária
def busca_binaria(lista, alvo):
    inicio = 0
    fim = len(lista) - 1
    while inicio <= fim:
        meio = inicio + (fim - inicio) // 2
        if lista[meio] < alvo:
            inicio = meio + 1
        elif lista[meio] > alvo:
            fim = meio - 1
        else:
            return meio  # Retorna o índice SE encontrar
    return -1  # Retorna -1 se não encontrar
# 2. Lista ordenada e entrada do usuário
numeros = [2, 3, 4, 10, 40]
alvo = int(input("Digite o número que deseja buscar: "))
# 3. Chamei a função e exibimos o resultado
resultado = busca_binaria(numeros, alvo)
if resultado != -1:
    print(f"O número {alvo} foi encontrado no índice {resultado}!")
else:
    print(f"O número {alvo} NÃO está na lista.")
```

**Explicação 21:** Com base na implementação de busca binária do GeeksforGeeks, criei a função **busca_binaria()** que recebe a lista e o alvo digitado no **input()**; Defini as variáveis **inicio** e **fim** para controlar os limites; Usei o **loop while** calculando o índice do **meio** com **inicio + (fim - inicio) // 2**; A cada iteração, ajusto os limites cortando a lista pela metade até retornar o índice com o **return** ou -**1** caso o número não exista.

**Exercício 22:** Crie uma função para criptografar uma mensagem usando a cifra de
César.

```python
# 1. Criei a função que criptografa a mensagem
def criptografar_cesar(texto, deslocamento):
    resultado = ""
    # 2. Percorri cada caractere do texto cifrado
    for caractere in texto:
        # Se for uma letra maiúscula
        if caractere.isupper():
           # ELe vai somar as letras (posições delas) para frente
            codigo = (ord(caractere) - ord('A') + deslocamento) % 26 + ord('A')
            resultado += chr(codigo)
        # Se for uma letra minúscula
        elif caractere.islower():
            # O mesmo processo só que com as minúsculas
            codigo = (ord(caractere) - ord('a') + deslocamento) % 26 + ord('a')
            resultado += chr(codigo)
        # Se for espaço ou pontuação, mantem original
        else:
            resultado += caractere
    return resultado
# 3. Entradas do usuário
mensagem= input("Digite a mensagem a ser criptografada: ")
chave = int(input("Digite o número do deslocamento usado: "))
# 4. Chamo a função e exibimos o resultado
mensagem = criptografar_cesar(mensagem, chave)
print(f"Mensagem criptografada: {mensagem}")
```

**Exercício 23:** Crie uma função para descriptografar uma mensagem cifrada com a cifra de César.

```python
# 1. Criei a função que descriptografa a mensagem
def descriptografar_cesar(texto_cifrado, deslocamento):
    resultado = ""
    # 2. Percorri cada caractere do texto cifrado
    for caractere in texto_cifrado:
        # Se for uma letra maiúscula
        if caractere.isupper():
            # Inverte o sinal, ao invés de somar, subtrai
            codigo = (ord(caractere) - ord('A') - deslocamento) % 26 + ord('A')
            resultado += chr(codigo)
        # Se for uma letra minúscula
        elif caractere.islower():
            # Inverte o sinal, ao invés de somar, subtrai de novo
            codigo = (ord(caractere) - ord('a') - deslocamento) % 26 + ord('a')
            resultado += chr(codigo)
        # Se for espaço ou pontuação, mantem original
        else:
            resultado += caractere
    return resultado
# 3. Entradas do usuário
mensagem_cifrada = input("Digite a mensagem criptografada: ")
chave = int(input("Digite o número do deslocamento usado: "))
# 4. Chamo a função e exibimos o resultado
mensagem_original = descriptografar_cesar(mensagem_cifrada, chave)
print(f"Mensagem Descriptografada: {mensagem_original}")
```

```python
# Fonte: https://www.dio.me/articles/como-fazer-uma-validacao-de-cpf-com-python
import re
def validar_cpf(cpf): # Aqui começa a criação da minha função (def)
    # 1. Extrai apenas os números do CPF digitado usando List Comprehension
    nums = [int(digito) for digito in cpf if digito.isdigit()]
    # 2. Rejeita se não tiver exatamente 11 números
    if len(nums) != 11:
        return False
    # 3. Rejeita CPFs com todos os dígitos iguais (ex: 111.111.111-11)
    if nums == [nums[0]] * 11:
        return False
    # 4. CÁLCULO DO 1º DÍGITO VERIFICADOR
    # Multiplica os 9 primeiros dígitos por pesos de 10 a 2 e soma
    soma_1 = sum(a * b for a, b in zip(nums[0:9], range(10, 1, -1)))
    digito_esperado_1 = (soma_1 * 10 % 11) % 10
    if nums[9] != digito_esperado_1:
        return False  # Primeiro dígito incorreto
    # 5. CÁLCULO DO 2º DÍGITO VERIFICADOR
    # Multiplica os 10 primeiros dígitos por pesos de 11 a 2 e soma
    soma_2 = sum(a * b for a, b in zip(nums[0:10], range(11, 1, -1)))
    digito_esperado_2 = (soma_2 * 10 % 11) % 10
    if nums[10] != digito_esperado_2:
        return False  # Segundo dígito incorreto
    # Se passou por todos os testes matemáticos, é válido
    return True
# --- PROGRAMA PRINCIPAL ---
cpf_usuario = input("Digite o CPF para validação (ex: 123.456.789-01 ou 12345678901): ")
# Verifica se o CPF foi digitado no formato com pontuação usando Regex (re.match)
formato_com_pontos = re.match(r'^\d{3}\.\d{3}\.\d{3}-\d{2}$', cpf_usuario)
if validar_cpf(cpf_usuario):
    if formato_com_pontos:
        print(f"O CPF {cpf_usuario} é VÁLIDO e está com a formatação correta!")
    else:
        print(f"O CPF {cpf_usuario} é VÁLIDO, só sem pontuação")
else:
    print(f"O CPF {cpf_usuario} é INVÁLIDO... Tente outro CPF...")
```

Unifiquei a lógica com a referência da DIO; Importei o módulo **re** para identificar o padrão de formatação com pontuação; Criei a lista numeros usando **List Comprehension** e **.isdigit()** para filtrar apenas os algarismos; Utilizei **zip()(para unir os números com seus pesos)** e **range()** com **sum()** para realizar a multiplicação ponderada dos dígitos verificadores; Apliquei a fórmula (soma * 10 % 11) **%** **10** para calcular e validar os dígitos $9$ e $10$ ;Exibi se o CPF é válido e se possui formatação adequada ou não.

**Exercício 31:** Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
# Fonte: https://www.w3schools.com/python/python_dsa_bubblesort.asp
# https://stackoverflow.com/questions/14836228/is-there-a-standardized-method-to-swap-two-variables-in-python
# 1 Criei a lista desordenada e peguei o seu tamanho com len()
lista = [64, 34, 25, 12, 22, 11, 90]
n = len(lista)
# 2 Criei o loop externo para controlar a quantidade de repetições necessárias
for i in range(n):
    # 3 Criei o loop interno para comparar os elementos vizinhos par a par
    for j in range(0, n - i - 1):
        # 4 Criei a verificação de troca de elementos do lugar se o da esquerda for maior
        if lista[j] > lista[j + 1]:
            # 5 Fiz a troca direta de posições em python (fonte 2)
            lista[j], lista[j + 1] = lista[j + 1], lista[j]
# 6 Exibi a lista em ordem na tela
print("Lista ordenada:", lista)
```

```python
# Fonte: https://www.hashtagtreinamentos.com/senhas-e-numeros-aleatorios-em-python
# 1 Importei a biblioteca random pra gerar elementos aleatórios
import random
# 2 Criei a função pra gerar a senha e recber  o tamanho desejado
def gera_senha(tamanho):
    # 3 Criei uma variável com todas as letras e números possíveis
    caracteres = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"
    senha = ""
    # 4 Fiz o loop for pra escolher os caracteres de forma aleatória
    for i in range(tamanho):
        # 5 O random.choice escolhe um caractere aleatório da lista
        senha += random.choice(caracteres)
    return senha
# 6 Pedi pro usuário o tamanho que ele deseja para a senha
tamanho_senha = int(input("Digite o tamanho da senha desejada: "))
# 7 Chamei a função e exibo a senha gerada na tela
senha_gerada = gera_senha(tamanho_senha)
print(f"Sua senha é: {senha_gerada}")
```

**Exercício 40:** Crie um programa para gerenciar uma lista de compras, permitindo adicionar, remover e listar itens.

```python
# Fonte: Medium / GitBook Gabriel Schade / Livro Nilo Menezes (2ª Edição)
# 1 Criei uma lista vazia para armazenar os itens de compras
lista_compras = []
# 2 Criei um loop while eterno para exibir o menu de opções continuamente
while True:
    print("\n--- MENU LISTA DE COMPRAS ---")
    print("1 - Adicionar item")
    print("2 - Remover item")
    print("3 - Listar itens")
    print("4 - Sair")
    # 3 Pedi pro usuário a opção desejada do menu
    opcao = input("Escolha uma opção (1-4): ")
    # 4 Adicionei um novo item na lista usando .append()
    if opcao == "1":
        item = input("Digite o nome do item para adicionar: ")
        lista_compras.append(item)
        print(f"'{item}' foi adicionado com sucesso!")
    # 5 Removi um item da lista usando .remove() depois de verificar se ele existe
    elif opcao == "2":
        item = input("Digite o nome do item para remover: ")
        if item in lista_compras:
            lista_compras.remove(item)
            print(f"'{item}' foi removido!")
        else:
            print("Esse item não está na sua lista!")
    # 6 Exibo todos os itens cadastrados na lista usando o loop for
    elif opcao == "3":
        if len(lista_compras) == 0:
            print("A sua lista de compras está vazia.")
        else:
            print("\nSua Lista de Compras:")
            for index, item in enumerate(lista_compras, start=1):
                print(f"{index}. {item}")
    # 7 Encerrei a execução do programa com o break
    elif opcao == "4":
        print("Saindo da Lista de Compras... Até logo!")
        break
    # 8 Criei umm texto caso as entradas do usuário sejam inválidas
    else:
        print("Opção inválida! Digite um número de 1 a 4.")
```

**Exercício 48:** Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
# Fonte: https://www.w3schools.com/python/python_dsa_insertionsort.asp
# 1 Criei uma lista com números desordenados
numeros = [12, 11, 13, 5, 6]
# 2 Passei pela lista do segundo elemento até o final com o for
for i in range(1, len(numeros)):
    chave = numeros[i]
    j = i - 1
    # 3 Usei o while para mover os elementos maiores que a chave uma posição para a direita
    while j >= 0 and numeros[j] > chave:
        numeros[j + 1] = numeros[j]
        j -= 1
    # 4 Encaixei a chave na posição correta
    numeros[j + 1] = chave
# 5 Exibi a lista ordenada na tela
print(f"Lista ordenada com Insertion Sort: {numeros}")
```

**Explicação 48:** Criei uma lista com números desordenados e usei o algoritmo **Insertion Sort**(*que empurra os maiores para a direita e os menores para a esquerda, coloca o número da vez no seu devido lugar*) para ordenar; Passei pela lista usando o **loop for** a partir do segundo elemento e guardei o número atual na variável chave. Em seguida, usei o **loop while** para comparar a chave com os elementos anteriores e mover os números maiores para a direita; Quando encontrei a posição correta, encaixei a chave e exibi a lista ordenada na tela.
